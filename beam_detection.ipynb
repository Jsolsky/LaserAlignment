```python
import requests
import base64
import numpy as np
from PIL import Image
import io
import matplotlib.pyplot as plt


response = requests.get("http://10.196.56.114:2000/image")
base64_data = response.json()
decoded_bytes = base64.b64decode(base64_data["base64"])

img_pil = Image.open(io.BytesIO(decoded_bytes))
img_array = np.array(img_pil)

# If the original array was multi-dimensional, you will need to reshape it
# nparr = nparr.reshape((original_shape_r, original_shape_c))

print(img_array.shape)
```

```python
plt.imshow(img_array[:,:,0])
plt.show()
plt.imshow(img_array[:,:,1])
plt.show()
plt.imshow(img_array[:,:,2])
plt.show()
print(img_array[:,:,0].var())
print(img_array[:,:,1].var())
print(img_array[:,:,2].var())
```
