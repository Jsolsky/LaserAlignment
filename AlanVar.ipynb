```python
# !pip3 install allantools
import allantools
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("laser_pos_x.csv")
df = df.iloc[-60*48:]
df.shape
```

```python
data = df['Value'].values
tau0 = 60  # seconds

(taus, adev, adev_err, adev_n) = allantools.oadev(data, rate=1.0/tau0, data_type='freq', taus='log10')

plt.loglog(taus, adev)
plt.grid(True, which='both', linestyle='-', linewidth=1)
plt.xlabel('Observation time tau (s)')
plt.ylabel('Allan Deviation')
plt.title("Laser Position X")
plt.show()

plt.plot(data)
plt.grid(True)
plt.show()
```
